## Block 2: Listen

Eine Liste speichert mehrere Werte in einer Variablen.
**Warum das hier wichtig ist:**  Wenn wir später `cursor.fetchall()` aufrufen, 
bekommen wir eine Liste von Datenbankzeilen zurück.

````python
# list mit eckigen Klammern anlegen
aerzte = ['Dr. Weber', 'Dr. Grosch', 'Dr. Klein', 'Dr. Dörrier', 'Dr. Bleidt']

# Ein Objekt vom Typen list enthält keines bis mehrere Elemente
# Länge der list
print(f'Anzahl Ärzte: {len(aerzte)}')

# Einzelne Elemente rufe ich mit ihrem Index auf (WICHTIG: Index beginnt bei 0!)
print(f'Erster Arzt: {aerzte[0]}')
print(f'Dritter Arzt: {aerzte[2]}')
print(f'Letzter Arzt: {aerzte[-1]}')
print(f'Vorletzter Arzt: {aerzte[-2]}')
````

````python
# Tuple, werden mit () runder Klammer angelegt und enthalten eine beliebige Anzahl an Werten
aerzte_2 = ('Dr. Bob', 'Dr. Grosch', 'Dr. Doom')
````

````python
for arzt in aerzte_2:
    print(f'    -> {arzt}')
````

Warum ist das so? Wir nutzen tuple überall dort, wo Werte **NICHT** verändert werden müssen bzw. sollen. Dadurch, dass die list Objekte jeder Zeit änderbar sind, sind sie 'dynamisch' und das verbraucht mehr Memory.

## Block 5: Exception Handling mit try/except

Programme müssen mit Fehlern (exceptions) umgehen können, ganz besonders wenn sie mit Datenbanken,
Netzwerken oder Dateien arbeiten.

**Warum ist das wichtig?** Wenn die Datenbankverbindung fehlschlägt (und das tut sie gerne mal),
soll das Skript nicht einfach abstürzen, sondern eine verständliche Fehlermeldung ausgeben und sauber beenden.

````python
# Ohne Exception Handling
# zahl = int('kein Text')   kann er nicht weil Value Error

# Nun mit exception Handling:

def zahl_eingeben(wert):
    try:
        ergebnis = int(wert)
        print(f'Zahl erkannt: {ergebnis}.')
        return ergebnis
    except ValueError:
        print(f'Fehler: "{wert}" ist keine gültige Zahl.')
        return None
````

````python
zahl_eingeben('42')
````

````python
zahl_eingeben('dsfewwefe')
````

````python
# So kann das in der Datenbankverbindung aussehen. Plus finally:


def verbindung_simulieren(passwort):
    print('Verbindung wird hergestellt......')
    try:
        if passwort != 'geheim':
            raise Exception('Authentifizierung fehlgeschlagen!')
        print('Verbindung erfolgreich!')
        print(' hier könnte unsere abfrage abgefeuert werden')
    except Exception as fehler:
        print(f'Fehler: {fehler}')
    finally:
        #finally läuft IMMER egal ob Fehler oder sauber durchgelaufen
        print('Verbindung wird geschlossen.')

print('----- Test 1: korrektes Passwort ----')
verbindung_simulieren('geheim')
print()
print('----- Test 2: falsches Passwort ----')
verbindung_simulieren('12345')
````

**Merke:** `finally` entspricht dem `conn.close()`, das wir gleich bei der echten Datenbankverbindung sehen werden.
Die Verbindung wird immer geschlossen, auch wenn ein Fehler aufgetreten ist!

## Block 6: Das with-Statement

Das `with`-Statement öffnet eine Ressource (Text-Datei, Bild-Datei, etc. pp.), nutzt sie, und schließt sie danach **automatisch**. Auch bei Fehlern.

````python
# with bei Dateizugriff
#Datei öffnen, schreiben, automatisch schließen:
with open('aerzte.txt', 'w', encoding='utf-8') as datei:
    datei.write('Dr. Weber - Kardiologie\n')
    datei.write('Dr. Klein - Chirurgie\n')
#Hier ist Datei dann automatisch geschlossen

#Datei wieder lesen:
with open('aerzte.txt', 'r', encoding='utf-8') as datei:
    inhalt = datei.read()

print(inhalt)
````

Bei der Datenbankverbindung sieht es ähnlich aus:
```python

# conn und cur sind hier dann automatisch geschlossen

## Block 8: Erste SELECT-Abfrag

Jetzt lesen wir echte Daten aus der Datenbank!

Das Muster ist hierbei immer dasselbe:

1. Verbindung öffnen
2. Cursor erstellen
3. SQL ausführen
4. Ergebnisse holen
5. Verbindung schließen

````python
import psycopg2

try:
    with psycopg2.connect(**VERBINDUNG) as conn:
        with conn.cursor() as cur:

            #SQL Abfrage ausführen:
            cur.execute('SELECT name, fachgebiet FROM personal.aerzte ORDER BY name;')

            #Alle Ergebnisse holen:
            aerzte = cur.fetchall()
    
    #Außerhalb des with- ´blocks, Verbindung wird geschlossen
    print(f' Gefunden: {len(aerzte)} Ärzte\n')

    for zeile in aerzte:
        # wir erinnern uns zeile ist ein Tuple: (name, fachgebiet)
        print(f'   {zeile[0]:<25} {zeile[1]}')

except psycopg2.OperationalError as fehler:
    print(f'Datenbankfehler: {fehler}')
````

## Block 9: DictCursor - Spalten per Namen ansprechen

Mit dem normalen Cursor kommen Zeilen als Tuple bei uns an: `zeile[0]`, `zeile[1]`, etc.

Mit `DictCursor` komen Zeilen als Dictionary raus: `zeile['name']`, ...

````python
import psycopg2
import psycopg2.extras # für DictCursor

try:
    with psycopg2.connect(**VERBINDUNG) as conn:
        # DictCursor statt normalen Cursor:
        with conn.cursor(cursor_factory=psycopg2.extras.DictCursor) as cur:
            
            cur.execute('SELECT name, fachgebiet, email FROM personal.aerzte ORDER BY name;')
            aerzte = cur.fetchall()
    print('Ärzte in der Klinik:')
    print('-' * 60)
    for arzt in aerzte:
        #Jetzt per column name!
        print(f'Name:             {arzt['name']}')
        print(f'Fachgebiet:       {arzt['fachgebiet']}')
        print(f'Email:            {arzt['email']}')
        print('-'*60)
except psycopg2.OperationalError as fehler:
    print(f'Datenbankfehler: {fehler}')
````

## Block 10: Parameterisierte Abfragen (Sicherheitsaspekt)

Niemals SQL-Abfragn mit String-Formatierung allein bauen!
Das öffnet Tür und Tor für **SQL-Injection-Angriffe**.

````python
#Bitte niemals so:
fachgebiet = 'Kardiologie'
# sql = f'SELECT * FROM personal.aerzte WHERE fachgebit = '{fachgebiet_suche}'"


# Richtig, parameterisieren mit %s:
fachgebiet = ('Kardiologie',)

try:
    with psycopg2.connect(**VERBINDUNG) as conn:
        # DictCursor statt normalen Cursor:
        with conn.cursor(cursor_factory=psycopg2.extras.DictCursor) as cur:

            # %s als Platzhalt, psycopg2 ersetzt ihn sicher:
            cur.execute(
                'SELECT name, email FROM personal.aerzte WHERE fachgebiet = %s;',
                fachgebiet
            )
            ergebnis = cur.fetchall()

    print(f'Kardiologen ({len(ergebnis)} gefunden):')
    for arzt in ergebnis:
        print(f'  {arzt['name']} - {arzt['email']}')

except psycopg2.OperationalError as fehler:
    print(f'Datenbankfehler: {fehler}')
````

## Block 11: Daten schreiben mit INSERT

Beim Schreiben von Daten gibt es einen wichtigen Unterschied zu SELECT:
NACH INSERT, UPDATE oder DELETE muss `conn.commit()` aufgerufen werden.
Sonst werden die Änderungen nicht gespeichert.

> Im `with`-Block von psycopg2 passiert das automatisch beim Verlassen und bei
> Fehlern wird automatisch `rollback()` aufgerufen.

````python
# Neuen Patienten anlegen:
neuer_patient = {
    'vorname' : 'Max',
    'nachname' : 'Mustermann',
    'email' : 'm.mustermann@mustermail.de',
    'geburtstag' : '1985-06-07',
    'versich' : 'Techniker Krankenkasse'
}

try:
    with psycopg2.connect(**VERBINDUNG) as conn:
        with conn.cursor() as cur:
            cur.execute(
                '''
                INSERT INTO patientendaten.patienten
                    (vorname, nachname, email, geburtsdatum, versicherung)
                VALUES (%s, %s, %s, %s, %s)
                RETURNING id;
                ''',
                (
                    neuer_patient['vorname'],
                    neuer_patient['nachname'],
                    neuer_patient['email'],
                    neuer_patient['geburtstag'],
                    neuer_patient['versich']
                )
            )
            #RETURNING id gibt die neue ID zurück:
            neue_id = cur.fetchone()[0]
        # with conn verlassen = automatischer commit!
    print(f'Patient angelegt mit ID: {neue_id}!')

except psycopg2.IntegrityError as fehler:
    print(f'Datenbank Constraint verletzt: {fehler}')
except psycopg2.OperationalError as fehler:
    print(f'Verbindungsfehler: {fehler}')
````

## Block 12: RLS in der Praxis erleben

Wir haben in der DB RLS eingerichtet.
Jetzt wir uns das ganze mal in Python an.

Wir verbinden uns einmal als `postgres` und einmal als `dr_weber` (sieht nur Diagnosen).

````python
def diagnosen_laden(benutzer, pw):
    """Lädt alle sichtbaren Diagnosen für den angegebenen User."""
    verbindung = {
        'host' : 'localhost',
        'dbname' : 'klinik',
        'user' : benutzer,
        'password' : pw,
        'port' : 5432
    }
    try:
        with psycopg2.connect(**verbindung) as conn:
            with conn.cursor(cursor_factory=psycopg2.extras.DictCursor) as cur:
                cur.execute(
                    'SELECT id, icd_code, beschreibung, erstellt_von '
                    'FROM patientendaten.diagnosen ORDER BY id;'
                )
                diagnosen = cur.fetchall()
            print(f'\nDiagnosen sichtbar für [{benutzer}]: {len(diagnosen)}')
            for d in diagnosen:
                print(f'  ID {d['id']}: {d['icd_code']} - Erstellt von: {d['erstellt_von']}')
            
    except psycopg2.OperationalError as fehler:
        print(f'Verbindungsfehler: {fehler}')
````

## Block 13: Alles zusammen

Wir schreiben jetzt eine Funktion `klinik_report()` die:

1. Alle Ärzte aus `personal.arzte`lädt
2. Für jeden Arzt soll die Anzahl seiner Patienten aus `patientendaten.patienten`gezählt werden
3. Es soll eine formatierte Übersicht sein

````python
def klinik_report():
    try:
        with psycopg2.connect(**VERBINDUNG) as conn:
            with conn.cursor(cursor_factory=psycopg2.extras.DictCursor) as cur:
                #Alle Ärzte laden
                cur.execute('SELECT id, name, fachgebiet FROM personal.aerzte;')
                aerzte =cur.fetchall()

                print('=== Klinik Report ===')
                print(f'Ärzte gesamt: {len(aerzte)}\n')

                for arzt in aerzte:
                    #Patienten pro Arzt zählen
                    cur.execute(
                        'SELECT COUNT(*) FROM patientendaten.patienten '
                        'WHERE behandelnder_arzt_id = %s',
                        (arzt['id'],)
                    )
                    anzahl = cur.fetchall()[0]

                    #Schreibe Ausgabe
                    print(f'{arzt['name']:<25} {arzt['fachgebiet']:<20} {anzahl} Patient(en)')

    except psycopg2.OperationalError as error:
        print(f'Fehler: {error}')


klinik_report()
````
