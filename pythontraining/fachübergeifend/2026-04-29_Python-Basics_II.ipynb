## While-Schleifen

Funktionsweise:

- Die Bedingung wird zuerst in der while-Schleife geprüft (im Kopf).
- Wenn die Bedingung `True` ist, werden die Anweisungen im Codeblock ausgeführt.
- Nach der Ausführung wird die Bedingung erneut geprüft.
- Die Scjhleife läuft so lange weiter, bis wir nicht mehr `True` zurückgegeben bekommen, sprich, wenn `False`, dann wird beendet. 

```python
count = 1

while count <= 5:
    print(count)
    count += 1
```

**ACHTUNG! Achtet bitte immer auf die Positionierung der Anweisungen im Schleifenblock!**

```python
count = 1

while count <= 5:
    count += 1    #Dadurch, dass jetzt count vor dem print() hochgezählt wird, bekommen  wir nicht mehr die Zahlen von 1-5, sondern von 2-6.
    print(count)
```

```python
#ACHTUNG BITTE NICHT AUSFÜHREN!

while True:
    input("Gib deinen Namen ein!")
```

**Man kann mit `while True` arbeiten, aber man muss dabei dringend darauf achten, dass es eine Abbruchbedingung gibt <br>
bzw. eine mit der Zeit erreicht wird.**

```python
while True:
    response = input("Gib deinen Namen ein! Schreibe \"Abbruch\" zum beenden.")
    if response.title() == "Abbruch":
        break
```

```python
counter = 0

while True:
    input("Schreib doch mal wieder.")
    counter += 1
    if counter == 5:
        break
```

```python
while True:
    password = input("Bitte geben Sie ein neues Passwort ein.")
    if len(password) >= 8 and password.isalnum():
        print("Das ist ein gutes Passwort!")
        break
    else:
        print("Passwort ist zu kurz oder nicht alphanummerisch!")
        continue
```

```python
count = 0

while count <= 10:
    count += 1
    if count == 5:
        continue
    if count == 8:
        break
    print(count)
```

## Flag

**Flags werden oft in Loops verwendet, um zu steuen, wann und durch welche Bedingungen diese beendet wird.**

```python
count = 5
flag = True

while flag:
    print(f'Countdown: {count}')
    count -= 1
    if count == 0:
        flag = False
```

```python
entries = [10, 20, 'thirty', 'thirty-five', 40, 50, 'about three-fiddy']
#entries_2 = [10, 20, 30, 35, 40, 50, 3.5]
error_flag = False

for entry in entries:
    if not isinstance(entry, (int, float)):
        error_flag = True
        print(f'Fehler: {entry} ist keine Zahl!')

if error_flag:
    print('Die Liste enthält ungültige Werte!')
else:
    print('Alle EInträge der Liste sind gültige Zahlen.')
```

```python
zahlen = [1,2,3,4,50,10]

gesuchte_zahl = int(input('Zahl eingeben:'))
zahl_gefunden = False

for z in zahlen:
    if gesuchte_zahl == z:
        zahl_gefunden = True
        break

if zahl_gefunden:
    print('Die Zahl wurde gefunden!')
else:
    print('Die Zahl wurde nicht gefunden!')
```

```python
# Zunächst haben wir eine Matrix, die nicht nur aus Integern, sondern auch aus einem String und einem Bool besteht
matrix_3 = [
    [1, "2", 3],
    [4, True, 5],
    [7, 8, 9]
]
```

```python
# Dann lassen wir uns einmal alle Elemente in der Matrix einzeln ausdrucken mit ihrem Index

for reihen_index, liste in enumerate(matrix_3):
    print(f'Zeile {reihen_index}:')
    for idx, zahl in enumerate(liste):
        print(f'\tIndex: {idx}, Wert: {zahl}')
```

```python
# Wir wollen nun, dass die Matrix nur Integer Werte enthält. Dazu wollen wir alle strings, floats und bools in int umwandeln

for liste in matrix_3:
    for zahl in liste:
        if isinstance(zahl, (str, float, bool)):
            int(zahl)
            print(zahl)
        else:
            print(zahl)

print(matrix_3)
# Das Problem bei diesem Ansatz ist nun, dass es zwar im print gewandelt (außer der Bool), aber nicht in-place in matrix_3
```

```python
# Wir müssen die Werte direkt auf den Listen umwandeln, dazu benötigen wir 

for reihen_index, liste in enumerate(matrix_3):
    print(f'Zeile {reihen_index}:')
    for idx, zahl in enumerate(liste):
        #das passiert jetzt auf jedem einzelnen Element innerhalb der drei listen Elemente der Matrix
        if isinstance(zahl, (str, bool, float)):  #hier wird geprüft, ob das Element kein Integer ist
            liste[idx] = int(zahl)   # Wenn ja, dann wird es über deinen Index, den wir durch enumerate() erhalten haben, direkt in siner liste angesteuert und der aktuelle Wert in einen Integer gecasted!
            print(f'\tIndex: {idx}, Wert: {liste[idx]}')
        else:
            print(f'\tIndex: {idx}, Wert: {zahl}')

print(matrix_3)
```

## Slicing

```python
zahlen = [0, 1, 2, 3, 4, 5, 6]
```

```python
print(zahlen[2:5])
```

**Denkt daran,der Anfang `[2` ist inklusiv, also eingeschlossen, aber das Ende `5]` ist exklusive, also ausgeschlossen!**

```python
x = zahlen[-1:2]
print(x)
type(x)
```

**Hier ergibt sich ein Problem. Python läuft von links nach rechts, der Endindex ist aber kleiner als der Startindex. Deshalb bekommen wir nur eine leere Liste zurück.**

```python
# Hier will ich die vier letzten Zahlen ohne die allerletze, wir müssen also immer von links nach rechts arbeiten

print(zahlen[-4:-1])
```

# Dictionaries

## Hinzufügen und Ändern von Einträgen
- Man kann in Dictionaries einfach neue Key-Value-Paare hinzufügen oder vorhandene Werte ändern

```python
student = {
    "name":"Tankred",
    "alter":35,
    "studiengang":"Wirtschafts-Informatik"
}
```

```python
#hinzufügen eine neuen key-value paares

student["semester"] = 6
print(student)
```

```python
#ändern eines vorhandenen Wertes

student["alter"] = 36
print(student)
```

**Entfernen von Einträgen**
- Man kann Einträge aus einem Dictionary mit `del` oder der Methode `.pop()` entfernen.

```python
#mit del
del student["semester"]
print(student)
```

```python
#entfernen mit .pop()
studiengang = student.pop("studiengang")
print(student)
print(studiengang)
```

```python
#Achtung dict ist keine list! Es hat nicht alle die selben Methoden!

student.remove("alter")
```

**Iteration über Dictionaries**

```python
# iteration über die keys

for key in student:
    print(key)
```

```python
for key in student:
    print(student[key])
```

Das funktioniert auch soweit alles ganz gut, aber wir haben hier dezidierte Methoden für.

```python
# die Methode .keys() gibt in der for Schleife die Keys aus.

for key in student.keys():
    print(key)
```

```python
# die Methode .values() gibt uns nur die Values aus

for value in student.values():
    print(value)
```

```python
# mit der methode .items() über die gesamten key-value-paare iterieren

for key, value in student.items():
    print(f'Schlüssel: {key}, Wert: {value}')
```

**Nützliche Methoden**
- `.keys()`: Gibt eine Liste aller Schlüssel im Dictionary aus.

```python
student.keys()
```

- `.values()`: Gibt eine Liste alle Werte im Dictionary zurück.

```python
student.values()
```

```python
student.items()
```

- `.update()`: Aktualisiert das Dictionary mit dem gegebenen Wert. Vorhandene key-value-Paare werden aktualisiert, neue werden hinzugefügt.

```python
student.update(alter = 37)
print(student)
```

```python
student.update({"id-number":123456})
print(student)
```

- `.get(key, default)`: Gibt den Wert für den angegebenen Key zurück, wenn der key nmicht existiert, wird der Wert hinter default zurückgegeben.

```python
#wofür das ganze?
student["adress"]
```

```python
student.get("adress")
```

```python
student.get("name")
```

```python
student.get("adress", "Es liegt kein solch ein Key vor!")
```
