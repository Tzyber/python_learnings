## Block2: Listen

Eine Liste speichert mehrere Werte in einer variablen. Sie ist geordnet und änderbar. Listen werden mit eckigen Klammern [] erstellt und die Werte werden durch Kommas getrennt.
**Warum das hier wichtig ist:** Wenn wir später 'cursor.fetchall()' aufrufen, bekommen wir eine Liste von Datenbankzellen zurück.

```
#Liste mit eckigen Klammern anlegen
aerzte = ["Dr. Smith", "Dr. Johnson", "Dr. Lee"]
patienten = ["Patient A", "Patient B"]

# Ein Objekt vom Typen List enthält keines bis unendlich viele Werte
leere_liste = []
print(f'Anzahl der Ärzte: {len(aerzte)}')
print(f'Anzahl der Patienten: {len(patienten)}')

# Einzelne Werte aus der Liste ausgeben
print(aerzte[0]) # Erster Arzt
print(aerzte[1]) # Zweiter Arzt
print(aerzte[-1]) # Immer letzter Arzt [-2] wäre der vorletzte Arzt
```

```
# for-loop: Alle Werte einer Liste durchlaufen
for arzt in aerzte:
    print(f'Hallo -> {arzt}!')
```

```
# tuple - unveränderliche Liste, mit runden Klammern
unveraenderliche_liste = ("Dr. Smith", "Dr. Johnson", "Dr. Lee")
print(unveraenderliche_liste)
# unveraenderliche_liste[0] = 'Dr. Mojang' # Das führt zu einem Fehler
raeume = ['OP', 'Aufwachraum', 'Pausenraum']
raeume_2 = ('OP', 'Aufwachraum', 'Pausenraum') # tuple

#man kan bei der list die elemente jeder zeit verändern/manipulieren
raeume[0] = 'Operations Saal'
print(raeume)
```

```
raeume_2[0] = 'Operations Saal' # Das führt zu einem Fehler, da tuple unveränderlich ist
```

Warum ist das so? In der Praxis verwenden wir Listen, wenn wir eine Sammlung von Daten haben, die sich ändern kann, wie z.B. die Patientenliste. Wir verwenden Tupel, wenn wir eine Sammlung von Daten haben, die sich nicht ändern soll, wie z.B. die Namen der Ärzte, die immer gleich bleiben. Nachteil an listen ist das sie mehr speicherplatz benötigen als tupel.

```
# liste von Tupeln oder so sehe auch fetchall() aus!
# jeder zeile in der Datenbank entspricht einem Tupel in der Liste
datenbank_ergebnis = [('lukas', 30), ('mark', 25), ('Mathias', 40)]

# Auf einzelne Spalten zugreifen
for patient in datenbank_ergebnis:
    name = patient[0] # Name ist die erste Spalte
    alter = patient[1] # Alter ist die zweite Spalte
    print(f'{name} ist {alter} Jahre alt.')
```

## Block 3: Dictionaries
Ein Dictionary speichert Daten in Schlüssel-Wert-Paaren. Es ist geordnet (ab Python 3.7) und änderbar. Dictionaries werden mit geschweiften Klammern {} erstellt und die Paare werden durch Kommas getrennt.
**Warum ist das wichtig?** Mit *DictCursor* in psycopg2 können wir die Ergebnisse unserer Datenbankabfrage als Dictionary zurückbekommen, was den Zugriff auf die Daten erleichtert, da wir die Spaltennamen als Schlüssel verwenden können. dies ist um einiges lesbarer

```
# Dictionaries werden mit geschweiften Klammern erstellt
arzt = {
    "name": "Dr. Smith",
    "fachgebiet": "Datenbankentwicklung",
    "alter": 45,
    "in_dienst": True
}

# Zugriff auf Werte über Schlüssel
print(arzt["name"])
print(arzt["fachgebiet"])
print(arzt["alter"])
print(arzt["in_dienst"])

# mit f-strings
print(f'{arzt["name"]} arbeitet in der {arzt["fachgebiet"]}')
```

```
# Lite von Dictionaries mit DirectCursor
datenbank_ergebnis = [
    {"name": "Lukas", "alter": 30},
    {"name": "Mark", "alter": 25},
    {"name": "Mathias", "alter": 40}
]

#zugriff per spaltenname mit f-strings
for patient in datenbank_ergebnis:
    name = patient["name"] # Name ist die erste Spalte
    alter = patient["alter"] # Alter ist die zweite Spalte
    print(f'{name} ist {alter} Jahre.')
```

## Block 4: Funktionen
Funktionen kapseln Code der mehrfach verwendet werden soll. Sie können Parameter entgegennehmen und Werte zurückgeben. Funktionen werden mit dem Schlüsselwort 'def' definiert.
**Warum ist das wichtig?** Funktionen helfen uns, unseren Code sauber und wiederverwendbar zu halten. Wenn wir z.B. eine Funktion haben, die eine Datenbankverbindung herstellt, können wir diese Funktion überall in unserem Code aufrufen, anstatt den Verbindungsaufbau jedes Mal neu zu schreiben.
`patienten_laden()`, `patienten_aktualisieren()`, `patienten_loeschen()` könnten Funktionen sein, die wir in unserem Datenbankprojekt verwenden.

```
# Einfache Funktion ohne Parameter
def klinik_info():
    print("Willkommen in der Klinik für Datenbankentwicklung!")

# Funktion aufrufen:
klinik_info()
```

```
# Funktion mit Parametern
def patienten_info(name, alter, arzt = None):
    print(f'Patientenname: {name}')
    print(f'Alter: {alter} Jahre')
    if arzt != 0:
        print(f'Zugewiesener Arzt: {arzt}')

#funktion aufrufen mit Argumenten
patienten_info("Dominik", 30, "Dr. dett")
```

```
# Funktion mit Rückgabewert
def berechne_alter_in_10_jahren(alter):
    return alter + 10
# Funktion aufrufen und Rückgabewert speichern
aktuelles_alter = 30
alter_in_10_jahren = berechne_alter_in_10_jahren(aktuelles_alter)
print(f'Dominiks Alter in 10 Jahren: {alter_in_10_jahren} Jahre')
```

```
# berechne alte rin zehn jahren mit f-string
def berechne_alter_in_10_jahren_f_string(alter):
    return f'Alter in 10 Jahren: {alter + 10} Jahre'
aktuelles_alter = 30

alter_in_10_jahren = berechne_alter_in_10_jahren_f_string(aktuelles_alter)
print(f'Dominiks {alter_in_10_jahren}')
```

wenn der `return-value` nicht in einer Variable oder in einem Datenkonstruktor aufgefangenwird, verwindet er auch nach aufruf der Funktionn wieder. man kann den aufgerufen `return-value` auch innerhalb der Funktionen verwenden.

## Block 5: Exception Handling mit try-except
Programme müssen mit fehlern (exceptions) umgehen können, ganz besonders wenn sie mit Datenbanken, Netzwerken oder Datein arbeiten.

**warum ist das wichtig?** Wenn die Datenverbindung fehlschlägt, soll das Skript nicht einfach abstürzen, sondern eine vrständliche Fehlermeldung ausgeben und sauber beenden. Mit try-except können wir Fehler abfangen und entsprechend darauf reagieren.

```
# Ohne Exception Handling
# zahl = int('kein text') # Das führt zu einem Fehler, da 'kein text' nicht in eine Zahl umgewandelt werden kann

# Mit exception Handling
def zahl_eingeben(wert):
    try:
        ergebnis = int(wert)
        print(f'Erkante Zahl lautet: {ergebnis}')
        return ergebnis
    except ValueError:
        print(f'{wert} ist keine gültige eingabe')
        return None

zahl_eingeben(3)
zahl_eingeben('zweiundfürzig')
```

```
# So kann das in der Datenbankverbindung aussehen

def verbindung_simulieren(password):
    print('verbindung wird hrgestellt.....')
    try:
        if password != 'geheim':
            raise Exception('Authentifizierung fehlgeschlagen!')
        print('Verbindung erfolgreich!')
    except Exception as e:
        print(f'Fehler: {e}')
    finally: # Finally entspricht dem conn.close. Die Verbindung wird immer geschlossen auch wenn ein fehler aufgetreten ist.
        print('Verbindungsversuch abgeschlossen.')


verbindung_simulieren('geheim1')
```

## Block 8: Erste SELECT-Abfrage

jetzt lesen wir echte Daten aus der Datenbank!
Das Muster ist heirbei das selbe:

1. Verbindung herstellen
2. Cursor erstellen
3. SQL-Abfrage ausführen
4. Ergebnisse verarbeiten
5. Verbindung schließen (oder mit with-Statement arbeiten)

```
import psycopg2
try:
    with psycopg2.connect(**VERBINDUNG) as conn:
        with conn.cursor() as cur:
            # SQL Abfrage ausführen
            cur.execute('SELECT name, fachgebiet FROM personal.aerzte ORDER BY name;')
            aerzte = cur.fetchall() # Alle Ergebnisse als Liste von Tupeln zurückbekommen

    print (f'gefunden: {len(aerzte)} Ärzte\n')
    for zeile in aerzte:
        print(f'Name: {zeile[0]}, Fachgebiet: {zeile[1]}')

except psycopg2.OperationalError as failure:
    print(f'Fehler bei der Verbindung: {failure}')
```

## Block 9: DirectCusor - Spalten per namen ansprechen

Mit deinen normalen Cursor kommen zeilen als Tuple bei uns an: `('Dr. Smith', 'Datenbankentwicklung')`, etc.
Mit `DirectCursor` kommen zeilen als Dictionary an: `{'name': 'Dr. Smith', 'fachgebiet': 'Datenbankentwicklung'}`, etc. Das macht den Zugriff auf die Spalten einfacher und lesbarer, da wir die Spaltennamen als Schlüssel verwenden können.

```

import psycopg2.extras # Für Directcursor

try:
    with psycopg2.connect(**VERBINDUNG) as conn:
        with conn.cursor(cursor_factory=psycopg2.extras.DictCursor) as cur:
             cur.execute('SELECT name, fachgebiet, email FROM personal.aerzte ORDER BY name;')
             aerzte = cur.fetchall()
    print (f'gefunden: {len(aerzte)} Ärzte\n')
    for zeile in aerzte:
        print(f'Name: {zeile["name"]}, Fachgebiet: {zeile["fachgebiet"]}, Email: {zeile["email"]}')
        print('-' * 85)
except psycopg2.OperationalError as failure:
    print(f'Fehler bei der Verbindung: {failure}')
```

## Block 10: parametrisierte Abfragen (Sicherheitsaspekt)

Niemals SQl-Abfragen mit String-Formatierung allein bauen!
Das öffnet Tür und Tor für **SQl-Injection-Angriffe**.

```
#Bitte niemals so machen:
fachgebiet = 'Kardiologie'
# sql = f"SELECT name FROM personal.aerzte WHERE fachgebiet = '{fachgebiet}';"

# Richtig, parametrisierte Abfrage mit %s:
fachgebiet =('Kardiologie',) # Parameter müssen als Tuple übergeben werden, auch wenn es nur ein Parameter ist
try:
    with psycopg2.connect(**VERBINDUNG) as conn:
        with conn.cursor(cursor_factory=psycopg2.extras.DictCursor) as cur:
            cur.execute('SELECT name FROM personal.aerzte WHERE fachgebiet = %s;', fachgebiet)
            aerzte = cur.fetchall()
        print(f'Kardiologen (len(aerzte)): {len(aerzte)}')
        for arut in aerzte:
            print(f'Name: {arut["name"]}')
except psycopg2.OperationalError as failure:
    print(f'Fehler bei der Verbindung: {failure}')
```

##Block 11: Daten schreiben mit INSERT

Beim schreiben von daten gibt es einen wichtigen Unterschied zu SELECT:
Nach INSERTm UPDATE oder DELETE muss `conn.commi()` aufgerufen werden.
Sonst werden die änderungen nciht gespeichert.

> Im 'with'- Block von psycopg2 passiert das automatisch beim verlassen und bei
> Fehlern wird automatisch `rollback()` aufgerufen.

```
# Neuen patienten anlegen:
neuer_patient = {
    'vorname' : 'Lukas',
    'nachname' : 'Mustermann',
    'email' : 'L.mustermann@webmail.com',
    'geburtstag' : '1985-06-05',
    'versich' : 'techniker Krankenkasse'
}

try:
    with psycopg2.connect(**VERBINDUNG) as conn:
        with conn.cursor() as cur:
            cur.execute(
                '''
                INSERT INTO patientendaten.patienten
                    (vorname, nachname, email, geburtsdatum, versicherung)
                VALUES (%s, %s, %s, %s, %s)
                RETURNING id;
                ''',
                (
                    neuer_patient['vorname'],
                    neuer_patient['nachname'],
                    neuer_patient['email'],
                    neuer_patient['geburtstag'],
                    neuer_patient['versich']
                )
            )

            neue_id = cur.fetchone()[0]

        print(f'Patient angelegt mit ID: {neue_id}!')

except psycopg2.IntegrityError as fehler:
    print(f'Datenbank Constraint verletzt: {fehler}')
except psycopg2.OperationalError as fehler:
    print(f'Allgemeiner Datenbankfehler: {fehler}')
```

## Block 12: RLS in der Praxis erleben

Wir haben in der DB RLS eingerichtet.
Jetzt sehen wir uns das ganze mal in Python an.

Wir verbinden uns einmal als `postgres` und einmal das `dr_weber` (sieht nur Diagnosen)

```
def diagnosen_laden(benutzer, pw):
    """Lädt alle sichtbaren Diagnosen für den angegebenen User."""
    verbindung = {
        'host': 'localhost',
        'dbname': 'klinik',
        'user': benutzer,
        'password': pw,
        'port': 5432
    }

    try:
       with psycopg2.connect(**verbindung) as conn:
        with conn.cursor(cursor_factory=psycopg2.extras.DictCursor) as cur:
            cur.execute("""
                SELECT id, icd_code, beschreibung, erstellt_von
                FROM patientendaten.diagnosen
                ORDER BY id;
            """)

            diagnosen = cur.fetchall()
            print(f'\nDiagnosen sichtbar für [{benutzer}]: {len(diagnosen)}')
            for diagnose in diagnosen:
                print(f"ID: {diagnose['id']}, ICD: {diagnose['icd_code']}, "
                      f"Beschreibung: {diagnose['beschreibung']}, "
                      f"Erstellt von: {diagnose['erstellt_von']}")

    except psycopg2.OperationalError as fehler:
        print(f'Verbindungsfehler: {fehler}')
    except psycopg2.ProgrammingError as fehler:
        print(f'SQL-Syntaxfehler: {fehler}')


# diagnosen_laden('postgres', 'Awb2tz')
diagnosen_laden('dr_weber', 'DrweberPass2025!')
```

## Block 13: Alles zusammen

Wir schrieben jetzt eine Funktion `klinik_report()` die:
1. Alle Ärzte aus personal.arzt lädt
2. Für jeden Arzt soll die Anzahl seiner Patienten aus `Patiendendaten.patienten`  gezählt werden.
3. Es soll eine formatiere Übersicht sein.

```
def klinik_report():
    try:
        with psycopg2.connect(**VERBINDUNG) as conn:
            with conn.cursor(cursor_factory=psycopg2.extras.DictCursor) as cur:
                cur.execute('SELECT id, name, fachgebiet FROM personal.aerzte ORDER BY name;')
                aerzte = cur.fetchall()

                print("=== Klinik Report ===")
                print(f'Ärzte gesamt: {len(aerzte)}\n')

                for arzt in aerzte:
                    # 2. Patienten für diesen Arzt zählen
                    cur.execute(
                        'SELECT COUNT(*) FROM patientendaten.patienten WHERE behandelnder_arzt_id = %s;',
                        (arzt['id'],)
                    )

                    anzahl = cur.fetchone()[0]

                    print(f"{arzt['name']:<25} {arzt['fachgebiet']:<25} {anzahl} Patient(en)")

    except psycopg2.Error as fehler:
        print(f'Datenbankfehler: {fehler}')

klinik_report()
```
